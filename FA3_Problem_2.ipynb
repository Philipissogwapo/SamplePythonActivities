```python
print("---Loan Eligibility Checker---\n")
print("This is the Loan Eligibility Checker, to use it, enter the following details to check if you're eligible for a loan.\n")
# This tells the user what the program is and shows them how to use it

while True:
    try:
        credit_score = int(input("Credit Score : "))
        annual_income = int(input("Annual Income (in dollar) : "))
        years_in_current_job = int(input("Years in your current job : "))
        break
    except ValueError:
        print("You cannot type that in, try again.\n")
# This block of code makes the user input the needed values. When they've successfully put the values, the block of code stops. When they enter a non integer, the except ValueError catches that and makes them do it again because of the while loop.

if credit_score < 700 or annual_income < 30000 or years_in_current_job < 2:
  print ("\nYou are not eligible for a loan. You have not met the specific requirement(s).")
else:
   print ("\nYou are eligible to receive a loan.")
# When they have a lacking requirement, the program tells the user that they cannot be eligible for a loan. If they have met the requirements of a loan, the program tells them that they are able to receive a loan.
```
