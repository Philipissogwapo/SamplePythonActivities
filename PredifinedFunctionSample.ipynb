```python
print ("---The Distance Calculator---")
print ("Welcome to the distance calculator. To start, enter the desired values of the two points.")
# These help the user understand what this is and what to do.

import math

while True:
  try:
    x1 = float(input("Please input a number for x1 : "))
    y1 = float(input("Please input a number for y1 : "))
    x2 = float(input("Please input a number for x2 : "))
    y2 = float(input("Please input a number for y2 : "))
    break
  except ValueError:
    print("Invalid input. Please enter numbers only.")
# This lets the user input what is needed. When they input a value that is not a float, the "except" makes them redo the whole thing.

distance = round(math.sqrt(pow( x2 - x1, 2 ) + pow( y2 - y1, 2 )), 2)
# This calculates the values to show the distance between the two points.

print(f"The distance between the two points is {distance}.")
# This shows the answer and lets the user know the answer.

# Using a library is more practical because it saves more time in doing calculations. A simple predefined function can take 5 seconds, while writing it from scratch can take minutes, maybe even hours. It also looks cleaner and better than a mess of code.
```
