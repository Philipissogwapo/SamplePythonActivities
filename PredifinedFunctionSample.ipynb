```python
import math

while True:
  try:
    x1 = float(input("Please input a number : "))
    y1 = float(input("Please input a number : "))
    x2 = float(input("Please input a number : "))
    y2 = float(input("Please input a number : "))
    break
  except ValueError:
    print("Invalid input. Please enter numbers only.")

x3 = (x2 - x1)
y3 = (y2 - y1)

xy = pow(x3, 2) + pow(y3, 2)
xxyy = math.sqrt(xy)

xxyyy = round(xxyy, 2)

print(f"The distance between the two points is {xxyyy}.")

# Using a library is more practical because it saves more time in doing calculations. A simple predefined function can take 5 seconds, while writing it from scratch can take minutes, maybe even hours. It also looks cleaner and better than a mess of code.
```
