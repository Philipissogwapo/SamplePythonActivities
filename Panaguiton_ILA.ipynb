```python
def fuel_efficiency(distance_traveled, fuel_consumed):
  fuel_efficiency = distance_traveled / fuel_consumed
  fuel_efficiency_round = round(fuel_efficiency, 1)
  return fuel_efficiency_round
# This is the function that takes the values and calculates the fuel efficiency by dividing the distance traveled by the fuel consumption.

print("---Fuel Efficiency Calculator---\n")
print("To use the fuel efficiency calculator, enter your distance traveled and the fuel consumption on the trip.\n")
# This introduces them to the program and teaches them how to use it.

while True:
  try:
    distance_traveled = float(input("Please enter your distance traveled (in kilometers) : "))

    if distance_traveled < 0:
      print("\nYou're not allowed to enter negative numbers. Try again.\n")
      continue
    else:
      break
  except ValueError:
    print("\nYou cannot enter that in. Try again.\n")
# This makes them input a positive float. If they put a negative number or a value that is not a number, it will ask them to do it again.

while True:
  try:
    fuel_consumed = float(input("\nPlease enter the fuel that was consumed (in liters) : "))

    if fuel_consumed < 0:
      print("\nYou're not allowed to enter negative numbers. Try again.")
      continue
    else:
      break
  except ValueError:
    print("\nYou cannot enter that in. Try again.")
# This is the same as the last one, they cannot enter a negative number or a value that is not a number.

fuel_efficiency = fuel_efficiency(distance_traveled, fuel_consumed)
# This just transfers the result to a variable so it can be used.

print(f"\nUsing the data you've given, your fuel efficiency is {fuel_efficiency} km/L.\n")
print("---End---")
# This shows the final result and that the program has ended.



```
