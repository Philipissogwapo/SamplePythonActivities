```python
print ("---Password Length Validator---\n")
print ("To use the Password Length Validator, you must enter a password that you would like, then the program will tell you if the password length is sufficient (8 - 15 characters).\n")
# This introduces the user to what the program is. It gives instructions on how to use it so the will not be confused.2423

while True:
  password = input("Please input your password : ")
  # This makes the user enter their desired password.

  password_length = len(password)
  # This checks the length of the password.

  if password_length < 8:
    print (f"Password too short! It only has {password_length} characters.\n")
  elif password_length > 15:
    print (f"Password too long! It has {password_length} characters.\n")
  else:
    print (f"Your password has enough characters. It has {password_length} characters.\n")
    break
  # This block of code compares the length to 8 and 15. If it's below 8 characters, it prompts the user to enter again because the length is too short. If it's above 15 characters, it prompts the user to try again because the length is too long.
```
